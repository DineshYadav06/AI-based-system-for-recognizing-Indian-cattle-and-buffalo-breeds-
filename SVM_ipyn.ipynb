```python
import os
import cv2
import numpy as np
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
from tqdm import tqdm
```

```python
# ---------------------------
# 1. Dataset Path (Your Path)
# ---------------------------
dataset_path = '/content/drive/MyDrive/Cattle_Breed_workspace/Kaggle_Datasets/cattle_breeds/PNG Breeds'
```

```python
# ---------------------------
# 2. HOG Feature Extractor
# ---------------------------
hog = cv2.HOGDescriptor()

def load_hog_dataset(folder_path, img_size=(128,128)):
    X = []
    y = []

    classes = sorted(os.listdir(folder_path))
    print("Classes found:", classes)

    for cls in classes:
        class_path = os.path.join(folder_path, cls)

        if not os.path.isdir(class_path):
            continue

        print(f"\nLoading Breed: {cls}")

        for img_name in tqdm(os.listdir(class_path)):
            img_path = os.path.join(class_path, img_name)

            img = cv2.imread(img_path)
            if img is None:
                continue

            img = cv2.resize(img, img_size)
            gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

            hog_features = hog.compute(gray)
            X.append(hog_features.flatten())
            y.append(cls)

    return np.array(X), np.array(y)
```
